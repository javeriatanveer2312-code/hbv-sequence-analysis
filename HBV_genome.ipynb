## 2. Clean the sequence
This step cleans the FASTA sequence retrieved from NCBI by removing line breaks and spaces.

```python
clean_dna = dna.replace("\n", "").replace(" ", "")
print(clean_dna)
```

## 3. Position of each base
Only the first 20 positions are printed here, to keep the output short.

```python
for i, bases in enumerate(clean_dna[:20]):
  print(f"position {i+1} : {bases}")
```

## 4. Codons
The sequence is read in groups of 3 bases. Only the first 20 codons are printed.

```python
for i in range(0, 60, 3):
  codon = clean_dna[i:i+3]
  print(codon)
```

## 5. First start codon (ATG)

```python
start_pos = clean_dna.find("ATG")
print(f"start_pos:{start_pos + 1}")
```

## 6. All start codons (ATG)

```python
count = 0
for i in range(len(clean_dna)):
  if clean_dna[i:i+3] == "ATG":
    count = count + 1
    print(f"Start codon #{count} found at biological position: {i+1}")
```

## 7. GC content

```python
g_count = clean_dna.count('G')
c_count = clean_dna.count('C')
total_gc = g_count + c_count
gc_percentage = (total_gc / len(clean_dna)) * 100
print(gc_percentage)
```
